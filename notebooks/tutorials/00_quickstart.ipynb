````python
from seqtrainer import DatasetRecipe, MaterializedDataset
from seqtrainer.transforms.dna import normalize_sequence, gc_content

print("Imports loaded")
````

## Create a tiny in-memory dataset

`MaterializedDataset` is a lightweight container for examples and metadata.

````python
examples = [
    {"sequence": "ATGCGTAA", "target": 0.91, "source": "demo_a"},
    {"sequence": "atnccgta", "target": 0.54, "source": "demo_b"},
    {"sequence": "GGGGTTTT", "target": 0.22, "source": "demo_c"},
]

dataset = MaterializedDataset(examples, metadata={"task": "promoter_regression"})
frame = dataset.to_pandas()
frame
````

## Basic sequence transforms

````python
frame["normalized"] = frame["sequence"].map(normalize_sequence)
frame["gc_content"] = frame["normalized"].map(gc_content)
frame
````

## Optional: define a recipe object

`DatasetRecipe` describes how data should be materialized from query results.

````python
recipe = DatasetRecipe(
    name="demo_recipe",
    query="SELECT ?sequence ?target WHERE { ... }",
    label_field="target",
    metadata_fields=("source",),
)

recipe
````
