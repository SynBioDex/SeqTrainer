# SeqTrainer Tutorial 01: Build a dataset from SBOL files

Use `seqtrainer.data.sbol` helpers to extract sequences and labels from local SBOL XML files.

```python
from pathlib import Path

import pandas as pd

from seqtrainer.data.sbol import build_dataset_from_files, get_sequence_from_sbol
```

```python
base = Path("/Users/gonzalovidal/Documents/GitHub/SeqTrainer/data/sbol_data")
sample_files = sorted(base.glob("sample_design_*.xml"))[:8]
len(sample_files), sample_files[0].name
```

```python
sample_files
```

## Inspect one file

```python
first_path = sample_files[0]
sequence = get_sequence_from_sbol(first_path)
print(first_path.name)
print(f"sequence length: {len(sequence) if sequence else 0}")
print(f"prefix: {sequence[:30] if sequence else None}")
```

## Build a tabular dataset

Rows where sequence or numeric target is missing are skipped.

```python
dataset_df = build_dataset_from_files(sample_files)
print(f"rows: {len(dataset_df)}")
dataset_df.head()
```

## Save for downstream modeling

```python
output_path = Path("data/dataset_builder/tutorial_dataset.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
dataset_df.to_csv(output_path, index=False)
output_path
```
