# SeqTrainer Tutorial 02: DNA feature engineering

Generate model-ready numeric features from raw sequences using `seqtrainer.transforms.dna`.

```python
import pandas as pd

from seqtrainer.transforms.dna import (
    gc_content,
    kmer_counts,
    normalize_sequence,
    one_hot_encode,
    pad_or_trim,
)
```

```python
sequences = [
    "ATGCATGC",
    "GGGGtttt",
    "NNNACGTX",
]

normalized = [normalize_sequence(s) for s in sequences]
normalized
```

## Standardize sequence length

```python
fixed_length = [pad_or_trim(s, length=10) for s in normalized]
fixed_length
```

## One-hot encoding tensor (`N x L x C`)

```python
encoded = one_hot_encode(fixed_length)
encoded.shape
```

## Scalar and dictionary features

```python
feature_rows = []
for seq in fixed_length:
    row = {"sequence": seq, "gc_content": gc_content(seq)}
    row.update({f"k2_{k}": v for k, v in kmer_counts(seq, k=2).items()})
    feature_rows.append(row)

features_df = pd.DataFrame(feature_rows)
features_df.iloc[:, :10].head()
```
