# SeqTrainer Tutorial 03: Train/val/test splits

Use `MaterializedDataset` to create reproducible train/validation/test splits.

```python
from seqtrainer.data.materialized import MaterializedDataset
```

```python
examples = [
    {"id": i, "sequence": "ATGC" * 5, "target": float(i) / 20.0}
    for i in range(20)
]

materialized = MaterializedDataset(examples, metadata={"origin": "tutorial"})
materialized.to_pandas().head()
```

```python
train_ds, val_ds, test_ds = materialized.train_val_test_split(
    train_size=0.7, val_size=0.15, test_size=0.15, seed=7
)

len(train_ds.examples), len(val_ds.examples), len(test_ds.examples)
```

```python
print(train_ds.metadata)
print(val_ds.metadata)
print(test_ds.metadata)
```
